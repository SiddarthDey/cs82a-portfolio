```python
temperatures = [28, 45, 62, 75, 90]
for temp in temperatures:
    if temp < 32:
        print(f"{temp}°F: Freezing")
    elif temp < 50:
        print(f"{temp}°F: Cold")
    elif temp < 70:
        print(f"{temp}°F: Mild")
    elif temp < 85:
        print(f"{temp}°F: Warm")
    else:
        print(f"{temp}°F: Hot")
```
