```python
def multiply_for(a, b):
    total = 0
    for _ in range(b):
        total += a
    return total

print(multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    total = 0
    count = 0
    while count < b:
        total += a
        count += 1
    return total

print(multiply_while(7, 6))
```
