```python
def pounds_to_kilograms(pounds):
    """Converts a weight in pounds to kilograms."""
    kg = pounds * 0.453592
    return kg
```

```python

weight_1 = 150
weight_2 = 200
weight_3 = 135

print(f"{weight_1} lbs is equal to {pounds_to_kilograms(weight_1):.2f} kg")
print(f"{weight_2} lbs is equal to {pounds_to_kilograms(weight_2):.2f} kg")
print(f"{weight_3} lbs is equal to {pounds_to_kilograms(weight_3):.2f} kg")
```
