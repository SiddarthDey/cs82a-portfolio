```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('custdata.tsv', sep='\t')


vars_of_interest = ['age', 'income', 'num.vehicles']

raw_corr = df[vars_of_interest].corr()
print("--- BEFORE CLEANING CORRELATION MATRIX ---")
print(raw_corr)
```

```python

print("--- RAW DATA SUMMARY ---")
print(df[vars_of_interest].describe())

initial_count = len(df)

cleaned_df = df[
    (df['age'] > 0) & 
    (df['age'] < 100) & 
    (df['income'] >= 0)
].dropna(subset=vars_of_interest)

cleaned_count = len(cleaned_df)
removed_count = initial_count - cleaned_count

print(f"\nInitial rows: {initial_count}")
print(f"Cleaned rows: {cleaned_count}")
print(f"Total rows removed: {removed_count}")
```

```python

cleaned_corr = cleaned_df[vars_of_interest].corr()
print("--- AFTER CLEANING CORRELATION MATRIX ---")
print(cleaned_corr)
```

Data Removal Summary & Justification

Total Rows Excluded: A total of records were removed during data cleaning.
Removal Criteria:
  1. Invalid/Impossible Age: Records with negative ages, zero age, or unrealistically high age entries (>100) were filtered out.
  2. Negative Income: Records with negative income values were removed as income cannot be below zero in valid survey reporting.
  3. Missing Data: Rows containing `NaN` / null values across `age`, `income`, or `num.vehicles` were dropped.
Why Cleaning Matters: Including invalid or uncleaned rows distorts the correlation coefficients $r$, producing deceptive relationships. Cleaning ensures the calculated correlation matrix accurately reflects real underlying trends.

Correlation Interpretation (Cleaned Data)

1. Age vs. Income: Shows a moderate positive correlation, indicating that income generally increases with age as individuals progress in their careers before stabilizing.
2. Income vs. Number of Vehicles: Displays a positive correlation, suggesting that higher-income households tend to own or lease more vehicles.
3. Age vs. Number of Vehicles: Shows a weak positive correlation, indicating a slight tendency for older individuals to own more vehicles compared to younger individuals.
