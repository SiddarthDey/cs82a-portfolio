```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('custdata.tsv', sep='\t')

initial_count = len(df)
print(f'Initial row count: {initial_count}')

print('Columns available:', df.columns.tolist())
```

```python

cleaned_df = df.dropna(subset=['housing.type'])
cleaned_df = cleaned_df[cleaned_df['housing.type'] != 'NA']


excluded_count = initial_count - len(cleaned_df)
print(f'Rows excluded: {excluded_count}')
print(cleaned_df['housing.type'].value_counts())
```

```python
housing_counts = cleaned_df['housing.type'].value_counts()

plt.figure(figsize=(8, 5))
sns.barplot(
    x=housing_counts.index,
    y=housing_counts.values,
    hue=housing_counts.index,
    legend=False,
    palette='viridis',
)
plt.title('Customer Distribution by Housing Type')
plt.xlabel('Housing Type')
plt.ylabel('Count / Number of Customers')
plt.xticks(rotation=30)
plt.show()
```

 Data Cleaning and Exclusion Summary

A total of 56 rows were excluded from the dataset due to missing or literal `"NA"` entries in the `housing.type` column. Removing these records ensures the final categorical bar chart accurately reflects only valid respondent data.
