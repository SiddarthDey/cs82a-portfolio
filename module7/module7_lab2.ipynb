```python
import pandas as pd

df = pd.read_csv('ratings.tsv', sep='\t')
print(df.head())
print(df.info())
```

```python
from sklearn.model_selection import train_test_split

X = df[['ambience', 'food', 'service']]
y = df['rating']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)

print("Intercept:", model.intercept_)
print("Test Score (R^2):", model.score(X_test, y_test))
```

```python
coefficients = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
}).sort_values(by='Coefficient', ascending=False)

print(coefficients)
```

Based on the multi-feature regression model, the coefficients ranked in order of impact on a restaurant's overall rating are:

Food (largest coefficient)

Service (moderate coefficient)

Ambience (smallest coefficient)

Recommendation for a struggling owner: A struggling restaurant owner should fix the food first. Since food carries the largest positive coefficient, improving food quality yields the highest expected increase in the overall rating compared to investing the same effort into ambience or service.
