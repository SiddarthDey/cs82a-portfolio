```python
import pandas as pd

df = pd.read_csv("truck.csv")
print(df.shape)
print(df.head())
```

```python
import matplotlib.pyplot as plt

correlation = df["X"].corr(df["Y"])
print("Pearson's r:", correlation)

ax = df.plot.scatter(x="X", y="Y", title="Truck List Price vs. Best Sale Price")
ax.set_xlabel("List Price (in thousands of dollars)")
ax.set_ylabel("Best Sale Price (in thousands of dollars)")
plt.show()
```

Prediction and Confidence Argument:
- Prediction: For a truck with a list price (X) of 18.0 (thousand dollars, or 18,000), we can predict a best sale price (Y) of approximately 16.1 (thousand dollars, or 16,100) based on the linear trend shown in the scatter plot.
- Confidence Argument: Because Pearson's r is approx 0.999, the relationship is exceptionally strong and linear, making our prediction highly reliable for list prices that fall within the observed range of this dataset (12 to 23 thousand dollars). However, this confidence does not extend to trucks priced outside this specific range, as extrapolating past the bounds of our data could lead to inaccurate estimates.
