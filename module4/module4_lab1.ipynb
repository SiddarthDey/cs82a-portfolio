```python
import pandas as pd

df = pd.read_csv("sales_clean.csv")
print(df.shape)
```

```python
print("--- Price Stats ---")
print("Mean:", df["price"].mean())
print("Median:", df["price"].median())
print("Std Dev:", df["price"].std())

print("\n--- Qty Stats ---")
print("Mean:", df["qty"].mean())
print("Median:", df["qty"].median())
print("Std Dev:", df["qty"].std())
```

For the price data, I would report the median (37.53) rather than the mean (55.72). Because the distribution is right-skewed with high-priced outliers pulling the mean upward, the median provides a much more honest and typical representation of item price.

By-Hand Calculation for [2, 4, 4, 6, 9]:

- Sum: 2 + 4 + 4 + 6 + 9 = 25
- Mean: 25 / 5 = 5.0
- Median The middle value of the sorted sequence is 4.
