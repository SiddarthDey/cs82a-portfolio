```python
import pandas as pd

df = pd.read_csv("allbacks.csv")
print(df.shape)
print(df.head())
```

```python

sorted_covers = df["cover"].sort_values().reset_index(drop=True)
median_cover = sorted_covers[len(sorted_covers) // 2]

print("--- Median Cover Type ---")
print(median_cover)


print("\n--- Overall Mean Weight ---")
print(df["weight"].mean())

print("\n--- Mean Weight by Cover Type ---")
print(df.groupby("cover")["weight"].mean())
```

Interpretations:
1. Median Cover Type: The median category for the book covers is paperback (`pb`), showing that at least half of the bookstore inventory consists of paperbacks when ordered alphabetically.
2. Overall Mean Weight: The average book in the inventory weighs roughly 706.67 grams.
3. Cover Type Comparison: Hardcover books (`hb`) are substantially heavier on average (about 796.43 grams) compared to paperbacks (`pb`), which average about 628.13 grams, indicating that cover binding type heavily influences total book weight.
