Plausibility Assumption for Bridge Lengths

Assumption: I assume that any bridge length exceeding 4,000 feet is implausible for this dataset and should be flagged as an outlier for manual inspection.

```python
import pandas as pd

df = pd.read_csv('bridges.csv')

print("Shape:", df.shape)
display(df.describe())
display(df.head())
```

```python

threshold = 4000
flagged_bridges = df[df['length'] > threshold]

print(f"Number of flagged bridges: {len(flagged_bridges)}")
display(flagged_bridges[['river', 'location', 'length']])
```

Handling Justification

- Flagged Value: Bridge at index 31 with a length of 4558.0.
- Action Taken: Inspected the value against the 4,000-foot plausibility assumption. Because it exceeds the threshold, it is flagged as an extreme outlier for potential verification or handling (such as treating it as missing or retaining it depending on whether historical verification confirms its accuracy).
