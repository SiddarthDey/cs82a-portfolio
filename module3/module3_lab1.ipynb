```python
import pandas as pd

df = pd.read_csv('messy_sales.csv')

print("Shape:", df.shape)
display(df.head(10))
print(df.dtypes)
```

Initial Observations
Missing Prices: The `price` column has missing values (`NaN`).
Date Formatting: The `date` column mixes different styles (like `MM/DD/YYYY` and `YYYY-MM-DD`).
Zip Code Lengths: Some zip codes are missing leading zeros (showing only 4 digits).

```python
df.isna().sum()
```

```python

fill_value = df['price'].median()
print("Fill value used (median):", fill_value)

df['price'] = df['price'].fillna(fill_value)

print("Missing prices remaining:", df['price'].isna().sum())
```

```python

print("Duplicates found:", df.duplicated().sum())

df = df.drop_duplicates()

print("New shape:", df.shape)
```

```python

df['zip'] = df['zip'].astype(str).str.zfill(5)

print(df['zip'].head(10))
```

```python

df['date'] = pd.to_datetime(df['date'], format='mixed')

print(df.dtypes)
```

```python

negative_qty = df[df['qty'] < 0]
display(negative_qty)
print("Count of negative quantity rows:", len(negative_qty))
```

```python

df = df[df['qty'] >= 0]

print("Shape after removing negative quantities:", df.shape)
```

```python

df.to_csv('sales_clean.csv', index=False)

print("File successfully saved as sales_clean.csv!")
```

Data Cleaning Log Summary
1. **Missing Prices**: Imputed 12 missing values in the `price` column using the median value ($37.53).
2. **Duplicates**: Removed 8 duplicate rows, reducing the total row count from 300 to 292.
3. **Zip Codes**: Converted zip codes to strings and padded them to 5 digits using leading zeros (e.g., fixing 4-digit zip codes).
4. **Dates**: Converted the `date` column into standard datetimes using mixed format parsing.
5. **Negative Quantities**: Filtered out 3 rows containing negative quantities (returns/anomalies), leaving a final cleaned shape of (289, 6).
6. **Export**: Saved the final processed dataset as `sales_clean.csv` without indices.
